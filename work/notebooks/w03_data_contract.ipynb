```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
import duckdb
from huggingface_hub import HfApi, hf_hub_download
from google.colab import userdata

# 1. Retrieve Hugging Face access token
HF_TOKEN = userdata.get('HF_TOKEN')
REPO_ID = "FlyRank/internship-warehouse"

# 2. Fetch repository file list
api = HfApi()
repo_files = api.list_repo_files(repo_id=REPO_ID, repo_type="dataset", token=HF_TOKEN)

# 3. Filter for all parquet partition files in the fact table directory
fact_files = [f for f in repo_files if f.startswith("fact_content_daily_performance/") and f.endswith(".parquet")]
print(f"Found {len(fact_files)} partition file(s) for the fact table.")

# 4. Download partition files to local cache
local_paths = [
    hf_hub_download(
        repo_id=REPO_ID,
        filename=f,
        repo_type="dataset",
        token=HF_TOKEN
    )
    for f in fact_files
]

# 5. Connect DuckDB and register view using formatted Python list string
con = duckdb.connect()
con.execute(f"CREATE VIEW fact_table AS SELECT * FROM read_parquet({local_paths})")

# 6. Verify total row count
total_rows = con.execute("SELECT COUNT(*) FROM fact_table").fetchone()[0]
print(f"Dataset successfully registered in DuckDB! Total rows: {total_rows:,}")

# 7. Inspect schema and column names
columns = [col[0] for col in con.execute("DESCRIBE fact_table").fetchall()]
print("Columns in fact table:", columns)
```
